### Magic Commands and Shell Escapes

**Magic** commands, are a component of IPython, the Jupyter kernel that allows you to use Python interactively within a Jupyter Notebook. Magic commands can help make your data analysis more efficient.
<br><br>
The syntax to call a magic function is:
`%` for line magics, and `%%` for cell magics.
<br><br>

Some examples:
- `%whos`: List every variable in kernel
- `%pip install package`: Install a package into the Python kernel
- `%time`: Time the execution of a line
- `%%time`: Time the execution of a cell
- `%timeit`: Time the execution of a line, averaged across 7 repeats 
- `%%timeit`: Time the execution of a line, averaged across 7 repeats 

_Note: Magic functions are part of the Python kernel and may not be available when using Jupyter Notebooks with other programming languages._ 

**Shell escapes**, which are a `!` prefix, send a command directly to your terminal, and return the result. 
- `!ls`: Print the contents of the directory (on Mac, Linux)
- `!dir`: Print the contents of a directory (on Windows)

### Interactive Widgets
`ipywidgets` lets you add sliders, dropdowns, and buttons that re-run code when changed.

````python
from ipywidgets import interact
@interact(n=(5, 20))
def plot_hist(n=10):
    sns.histplot(data=penguins_df, x="flipper_length_mm", bins=n)
    plt.show()
````

````python
columns = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]

@interact(
    column=columns,
    bins=(5, 40),
    species=["All", "Adelie", "Chinstrap", "Gentoo"],
)
def explore(column="flipper_length_mm", bins=15, species="All"):
    if species == "All":
        data = penguins_df
    else:
        data = penguins_df[penguins_df["species"] == species]

    sns.histplot(data=data, x=column, bins=bins)
    plt.title(f"{column} ({species}, n={len(data)})")
    plt.show()
````
